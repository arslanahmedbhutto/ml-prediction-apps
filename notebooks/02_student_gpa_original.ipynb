```python
#Feature Engineering (har koi nh kr sakta (need domain expert))

x = df.drop(["GPA","GradeClass"],axis=1) #independent variables
```

```python
x.head()
```

```python
y= df["GPA"] #depend (target variable)
```

```python
#model
from sklearn.model_selection import train_test_split

x_train,x_test,y_train, y_test = train_test_split(
    x,y,test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_squared_error, r2_score

mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Squared Error:", mse)
print("R2 Score:", r2)
```

```python
import joblib

joblib.dump(model,"/content/drive/MyDrive/AI_tasks/student_model.pkl")
```
